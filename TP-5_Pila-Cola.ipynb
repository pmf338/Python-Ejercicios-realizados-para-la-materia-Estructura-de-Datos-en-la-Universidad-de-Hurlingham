### **Ejercicio 1**

Implementar el TDA Pila (Stack), con las siguientes operaciones:
- Crear (__init__())
- Vaciar
- Apilar elemento (push)
- Desapilar elemento (pop)
- Obtener primer elemento (top)
- Obtener tamaño de pila
- Clonar
- Esta vacía.
- __repr__(). Para poder imprimir una Pila por consola

```
import copy as cp # paquete para no tener problemas con las estructuras que tienen multiples niveles

class Pila:
  def __init__(self, listaInicial=None):
    self.pila = []
    if listaInicial:
      self.pila = cp.deepcopy(listaInicial)

  def tamaño(self)->int:
    return len(self.pila)

  def estaVacia(self)->bool:
    return self.tamaño() == 0

  def apilar(self, dato:int)->None:
    self.pila.append(dato)

  def tope(self)->int:
    datoCima = None
    if not self.estaVacia():
      datoCima = self.pila[len(self.pila)-1] # leo ultimo elemento
    return datoCima

  def desapilar(self)->int:
    dato = None
    if not self.estaVacia():
      dato = self.pila.pop()
    return dato

  def vaciar(self)->None:
    self.pila = []

  def clonar(self): # retorna una pila que contengan los mismos elementos pero en otra posicion de memoria
    return cp.deepcopy(self)

  def __repr__(self)->str:
    return str(self.pila)+"<->"
```

```
pila1 = Pila()
print(pila1)
pila1.apilar(3);pila1.apilar(2);pila1.apilar(7);pila1.apilar(4)
print(pila1)
print(pila1.tope())
print(pila1.desapilar())
print(pila1)
print(pila1.tope())
pila2 = pila1.clonar()
print("pila2",pila2)
```

### **Ejercicio 3**

Escribir una función que invierta el orden de una pila. No debe devolver una nueva pila invertida, sino invertir la pila que ingresa por parámetro.

```
def invertir(pila:Pila)->None: #declaro una función que tome una pila de tipo Pila y le invierte el orden
  listaAux = [] #Inicializo una lista auxiliar vacía. Es auxiliar porque no la vamos a retornar. Existe solo como paso intermedio.
  while not pila.estaVacia(): #Mientras pila NO esté vacía:
    listaAux.append(pila.desapilar()) # agrego en lista auxiliar: pila.desapilar(), que no solo borra el último elemento de la pila sino también que lo devuelve. En la misma línea asignamos un nuevo elemento en listaAux y quitamos el último de pila
  for dato in listaAux: #Ahora, para cada dato en mi lista auxiliar:
    pila.apilar(dato) #Apilamos nuevamente en pila los datos que tenemos guardados en listaAux. De esta forma no retornamos nada porque modificamos la pila original, vaciandola y volviéndola a llenar invertida

pila1 = Pila([2,3,1,5,6])
print(pila1)
invertir(pila1)
print(pila1)
```

```
#Otra forma de verlo
def invertir2(pila:Pila)->None: #Defino mi función, igual que antes
  pilaAux = pila.clonar() #En este caso como estructura auxiliar en vez de usar una lista uso la misma pila de entrada, clonada
  pila.vaciar() #Como ya tengo un clon, voy a vaciar mi pila original
  while not pilaAux.estaVacia(): #Mientras la pila auxiliar NO esté vacía:
    pila.apilar(pilaAux.desapilar()) #Voy a apilar en mi pila original (que está vacía) los elementos que voy desapilando de mi pila auxiliar

pila1 = Pila([2,3,1,5,6])
print(pila1)
invertir2(pila1)
print(pila1)
```

### **Ejercicio 6**

Escribir una función que elimine de una pila todas las ocurrencias de un elemento dado. Usar una pila auxiliar.

```
def eliminarElemento(unaPila:Pila,elementoAEliminar:int) -> None:
  pilaAux = Pila()
  while(not unaPila.estaVacia()):
    elemento = unaPila.desapilar()
    if(elemento != elementoAEliminar):
      pilaAux.apilar(elemento)
  while(not pilaAux.estaVacia()):
    unaPila.apilar(pilaAux.desapilar())

miPila = Pila([1 , 2 , 3 , 4 , 2])
eliminarElemento(miPila,2)
print(miPila)
```

La función debe retornar una nueva pila con el resultado.

Por ejemplo para sumar: 85699625 + 75426652

Las pilas de entrada a la función son:

Pila1 = 8, 5, 6, 9, 9, 6, 2, 5

Pila2 = 7, 5, 4, 2, 6, 6, 5, 2

La salida:

pilaSalida = 1, 6, 1, 1, 2, 6, 2, 7, 7

```
def sumaDePilas(p1:Pila,p2:Pila)->Pila:
  pilaSalida = Pila()
  sobrante = 0 # es la decima sobrante de la suma(se lleva siempre 1 porque las sumas de unidades nunca llegan a 20)

  while not p1.estaVacia(): # como len(p1) == len(p2), repite mientras p1 no este vacia
    if (p1.tope() + p2.tope() + sobrante) >= 10: # si la suma es >= 10, apila la unidad(suma - 10) y se guarda la decena(1)
      pilaSalida.apilar(p1.desapilar() + p2.desapilar() + sobrante - 10)
      sobrante = 1
    else: # caso contrario, suma todo y deja 0 de sobrante
      pilaSalida.apilar(p1.desapilar() + p2.desapilar() + sobrante)
      sobrante = 0
  pilaSalida.apilar(sobrante)
  invertir(pilaSalida)
  return pilaSalida
```

### **Ejercicio 11**

Implementar el TDA Cola (Queue), con las siguientes operaciones:
- Crear (__init__())
- Vaciar
- Encolar elemento (enqueue)
- Desancolar elemento (dequeue)
- Obtener primer elemento (top)
- Obtener tamaño de cola
- Clonar
- Esta vacía.
- __repr__(). Para poder imprimir una Cola por consola

```
class Cola:
  def __init__(self):
    self.cola = []

  def tamaño(self)->int:
    return len(self.cola)

  def estaVacia(self)->bool:
    return self.tamaño() == 0

  def encolar(self, dato:int)->None:
    self.cola.append(dato) #Encolando por el final
    #self.cola.insert(0,dato) #Encolando por el inicio

  def tope(self)->int:
    datoCima = None
    if not self.estaVacia():
      datoCima = self.cola[0]  #Desencolo por el inicio
      #datoCima = self.cola[len(self.cola)-1] #Desencolo por el final
    return datoCima

  def desencolar(self)->int:
    dato = None
    if not self.estaVacia():
      dato = self.cola.pop(0)  #Desencolando por el inicio
      #dato = self.cola.pop() #Desencolando por el final
    return dato

  def vaciar(self)->None:
    self.cola = []

  def clonar(self):
    return cp.deepcopy(self)

  def __repr__(self)->str:
    return "<-"+str(self.cola)+"<-"
```

```
#colaOriginal = <-[2,3,1,5]<-
#cola = <-[5,1,3,2]<-
#pilaAux = []<->
def invertirCola(unaCola:Cola) -> None:
  pilaAux = Pila()
  while(not unaCola.estaVacia()):
    pilaAux.apilar(unaCola.desencolar())
  while(not pilaAux.estaVacia()):
    unaCola.encolar(pilaAux.desapilar())

cola = Cola()
cola.encolar(2)
cola.encolar(3)
cola.encolar(1)
cola.encolar(5)
invertirCola(cola)
print(cola)
```

### **Ejercicio 16**

Escribir una función que recibe una cola de enteros y genera dos colas: una con los números primos y otra con los números no primos, de la cola de entrada. Al terminar la función, la cola de entrada no debe estar modificada.

```
def esPrimo(numero:int) -> bool:
  resultado = True
  if(numero < 2):
    resultado = False
  else:
    for i in range(2,numero):
      if(numero % i == 0):
        resultado = False
  return resultado

def generarDosColas(unaCola:Cola) -> None:
  colaPrimos = Cola()
  colaNoPrimos = Cola()
  while(not unaCola.estaVacia()):
    if(esPrimo(unaCola.tope())):
      colaPrimos.encolar(unaCola.desencolar())
    else:
      colaNoPrimos.encolar(unaCola.desencolar())
  return colaNoPrimos

cola = Cola()
cola.encolar(2)
cola.encolar(3)
cola.encolar(1)
cola.encolar(5)
print(generarDosColas(cola))
```
