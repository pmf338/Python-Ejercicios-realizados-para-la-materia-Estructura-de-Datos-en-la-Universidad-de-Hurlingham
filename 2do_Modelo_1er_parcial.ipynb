Por último implementar la función  **masParesQue10(arreglo)**, que devuelve True cuando “arreglo” tiene más números pares que números mayores que 10. Esta última función no hace falta que sea recursiva. Las soluciones que NO sean recursivas no serán tomadas en cuenta.

Por ejemplo:

Si arr1 = [5, 10, 9, 8, 13, 21] => masParesQue10(arr1) devuelve False, porque arr1 tiene 2 números mayores que 10 y 2 números pares.

Si arr2 = [5, 10, 24, 9, 8, 6, 21] => masParesQue10(arr2) devuelve True, porque arr2 tiene 4 números pares y solamente 2 números mayores que 10.

Ayuda: Se puede asumir que ya se tiene la función esPar(n) que devuelve verdadero cuando n es par.


```python
import numpy as np

def cantidadPares(vector:np.array) -> int:
  cantidad = 0
  if(len(vector) == 0):
    cantidad = 0
  elif(vector[0] % 2 == 0):
    cantidad = cantidadPares(vector[1:]) + 1
  else:
    cantidad = cantidadPares(vector[1:])

  return cantidad

def cantidadMayoresADiez(vector:np.array) -> int:
  cantidad = 0
  if(len(vector) == 0):
    cantidad = 0
  elif(vector[0] > 10):
    cantidad = cantidadMayoresADiez(vector[1:]) + 1
  else:
    cantidad = cantidadMayoresADiez(vector[1:])

  return cantidad

def masParesQueDiez(vector:np.array) -> int:
  return cantidadPares(vector) > cantidadMayoresADiez(vector)


vector = [4, 10, 24, 9, 8, 6, 21]
print(cantidadPares(vector))
```

# Ejercicio 2

En UNAHUR se está desarrollando un software para administrar el estacionamiento. En todo momento pueden entrar o salir autos. No puede haber dos autos en el mismo lugar y de cada auto solo nos interesa conocer la patente y la hora de ingreso, que a fines prácticos consideraremos dos números.

Al inicio del día se puede asumir que el estacionamiento comienza vacío y con una cantidad fija de lugares que informa el bedel al abrir el lugar. Vamos a suponer que el estacionamiento tiene igual cantidad de espacios verticales que horizontales cada día. Por ejemplo 20x20 lugares. Lo que se necesita hacer es asignar un auto a algún lugar vacío cada vez que ingresa y desasignarlo cuando egresa.

Implementar el TDA EstacionamientoUnaHur que contenga al menos las siguientes operaciones:

* **__init__(cantLugares):** Recibe la cantidad total de lugares en el estacionamiento y crea una variable de tipo EstacionamientoUnaHur.

* **estacionar(patente, horaIngreso):** Estaciona el auto en algún lugar disponible. La función debe retornar la ubicación (posición vertical y posición horizontal) utilizada. Si no hay más lugar debe lanzar una excepción. Se puede asumir que no se estacionan dos autos iguales ni estacionan dos autos al mismo tiempo.

* **salir(patente):** Libera el espacio si el auto se encuentra en el estacionamiento y retorna la hora de ingreso. Si el auto no se encuentra en el estacionamiento no debe hacer nada.

* **estaVacio():** Retorna True si no hay ningún auto estacionado y False en caso contrario.

* **cantidadAutosHora(hora):** Devuelve la cantidad de autos que ingresaron antes de la hora recibida por parámetro y que aún están en el estacionamiento.

* **__repr__():** que muestra los lugares del estacionamiento.

Ayuda:

Para retornar dos valores (x,y) se puede hacer return x,y
Generar TDA auto antes del TDA estacionamiento.

**Notas:**

Agregar las operaciones y funciones que consideren, además de las pedidas.
Un estacionamiento vacío contiene None o 0 en todas las posiciones.
No se puede linealizar el estacionamiento (es decir, hacerlo de una dimensión)
Se debe crear un auto antes de asignarlo al lugar de estacionamiento.


```python
def validarTipo(variable:any, nombre:str, tipo:type)->any:
  if not isinstance(variable, tipo):
    raise Exception(f"La variable {nombre} debe ser de tipo {tipo}.")
  return variable

#Para poder hacer esto se deben usar punteros a funcion, sino con el tipo incorrecto la condicion se evalua antes
#de llamar a la funcion y va a fallar
def validarTipoCondicion(variable:any, nombre:str, tipo:type, condicion:bool = True):
  validarTipo(variable, nombre, tipo)
  if not condicion:
    raise Exception(f"La variable {nombre} no cumple la condición.")
  return variable

def validarNumeroPositivo(numero:int, nombre:str, tipo:type)->int:
  validarTipo(numero, nombre, tipo)
  if numero < 0:
    raise Exception(f"La variable {nombre} debe contener un numero positivo.")
  return numero
```

```python
import numpy as np

class Auto:
  def __init__(self,patente:int,horaIngreso:int):
    self.__patente = validarNumeroPositivo(patente,"patente",int)
    self.__horaIngreso = validarNumeroPositivo(horaIngreso,"horaIngreso",int)

  def __repr__(self) -> str:
    return f"{self.__patente} {self.__horaIngreso}"

  def esPatenteBuscada(self,patente) -> bool:
    return self.__patente == patente

  def getHoraIngreso(self) -> int:
    return self.__horaIngreso

class EstacionamientoUnaHur:
  def __init__(self,cantLugares:int):
    self.__cantLugares = validarNumeroPositivo(cantLugares,"cantLugares",int)
    self.__estacionamiento = np.empty((cantLugares//2,cantLugares//2),Auto)

  def __repr__(self) -> str:
    return f"{self.__estacionamiento}"

  def estacionar(self,patente, horaIngreso) -> int:
    asignado = False
    auto = Auto(patente,horaIngreso)
    for fila in range(self.__cantLugares//2):
      for columna in range(self.__cantLugares//2):
        if(self.__estacionamiento[fila][columna] == None):
          self.__estacionamiento[fila][columna] = auto
          asignado = True
          return (fila,columna)

    raise Exception(f"No hay lugar disponible")

  def salir(self,patente):
    liberado = False
    for fila in self.__estacionamiento:
      for auto in fila:
        if(auto != None and auto.esPatenteBuscada(patente)):
          horaDeIngreso = auto.getHoraIngreso()
          auto = None       # NO ME LO VACIA
          liberado = True
          return horaDeIngreso

  def estaVacio(self) -> bool:
    estacionamientoVacio = True
    for fila in self.__estacionamiento:
      for auto in fila:
        if(auto != None):
          estacionamientoVacio = False

    return estacionamientoVacio

  def cantidadDeAutosHora(self,hora) -> int:
    cant = 0
    for fila in self.__estacionamiento:
      for auto in fila:
        if(auto.getHoraIngreso() < hora):
          cant += 1

    return cant
```

```python
#El TDA EstacionamientoUnaHur deberá superar este test:
# 2x2 = 4 lugares de capacidad

print("Declaro mi estacionamiento:")
estacionamiento = EstacionamientoUnaHur(4)
print(estacionamiento, "\n")

print("Estaciono dos autos")
estacionamiento.estacionar(3333, 4)  # patente, hora
estacionamiento.estacionar(3334, 4)
print(estacionamiento,"\n")

print("Libero un lugar en el estacionamiento (sale el auto con patente 3333)")
estacionamiento.salir(3333)
print(estacionamiento, "\n")
print(estacionamiento.salir(3333))

print("Estaciono otros dos autos")
estacionamiento.estacionar(33335, 5)
estacionamiento.estacionar(33336, 5)
print(estacionamiento, "\n") #ahora hay 3 autos estacionados


#hay un auto que estaciono antes de las 5 (patente 3334) que aún está en el estacionamiento -> debe imprimir 1
print("Cantidad de autos estacionados antes de las 5: ", estacionamiento.cantidadDeAutosHora(5))
```
