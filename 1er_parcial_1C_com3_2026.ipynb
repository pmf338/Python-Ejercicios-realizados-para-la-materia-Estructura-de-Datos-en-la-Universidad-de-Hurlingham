# Ejercicio 1:
Escribir la función **esRegresiónAritmetica(vector: np.array) → bool**, que recibe un vector de enteros y retorna True si la
diferencia entre cada elemento y su siguiente es siempre la misma y menor . Si el vector de entrada está vacío o tiene un solo
elemento la función debe retornar False. Soluciones que **NO** sean recursivas **NO** serán tomadas en cuenta.

**Nota**: La definición de la función puede ser modificada, agregando parámetros de ser necesario. Si se resuelve usando varias
funciones, al menos una debe ser recursiva.

Por ejemplo:

v1 = [14, 11, 8, 5, 2] => esRegresiónAritmetica(v1) → True ( diferencia siempre 3 y sus siguientes menores)

v2 = v2 = [14, 11, 8, 5, 6] => esRegresiónAritmetica(v2) → False ( el anteultimo elemento y el ultimo no cumple con la misma diferencia)  

v3 = [20, 16, **12**, **11**, 7, 3] => esRegresiónAritmetica(v3) → False ( 11 y 12 no cumplen la diferencia de 4)

```python
import numpy as np

def esRegresionAritmetica(vector: np.array) -> bool:
  if(len(vector) <= 1):
    resultado = False
  else:
    diferencia = vector[0] - vector[1]
    resultado = calculoRegresionAritmetica(vector, diferencia)

  return resultado


def calculoRegresionAritmetica(vector: np.array, diferencia) -> bool:
  if(len(vector) <= 1):
    resultado = True
  elif(vector[0] <= vector[1] or vector[0] - vector[1] != diferencia):
    resultado = False
  else:
    resultado = calculoRegresionAritmetica(vector[1:], diferencia)

  return resultado

vector = [20, 16, 12, 11, 7, 3]
print(esRegresionAritmetica(vector))
```

# Ejercicio 2
Aeropuertos Argentina decide desarrollar un software para optimizar el funcionamiento del aeropuerto. Implementar el TDA
Pista que modela una pista de un aeropuerto, con N filas con X posiciones cada una. En cada posición se ubicará un Vuelo, que
se define con el nombre del destino, la matrícula del avión, cantidad de horas en posición y el origen, que puede ser:
“nacional” o “internacional”.

Al ubicar vuelos en la pista, tendremos en cuenta el origen, a vuelos “internacionales” los tenemos que ubicar en la posiciones
que dan a migraciones (primera y segunda fila), mientras que los vuelos con origen “nacional” solo pueden ir en el resto de la
pista.

Nota: Implementar el TDA Vuelo con las operaciones necesarias. Pueden suponer que la cantidad de filas siempre es mayor o
igual a 3 (al menos 2 filas de vuelos internacionales y 1 para nacionales).

Implementar el TDA **Pista** conteniendo al menos las siguientes operaciones:

- **__init__(self, cantFilas:int, cantPosFila:int)**: que recibe la cantidad de filas y la cantidad de posiciones por fila y crea una variable de tipo Pista.

- **arriboDelVuelo(self, nombre:str, mat:str, cantHorasPos:int, origen:str)**: que recibe los datos de un vuelo
y lo ubica en la pista. Arroja una excepción si el vuelo no puede ser ubicado, por ejemplo si es un vuelo internacional
y las posiciones correspondientes están ocupadas.

- **esArriboPernocte(self) → bool**: que retorna True si encuentra al menos un vuelo que la cantidad de horas en posición sea mayor o
igual a 6.

- **cantPernocte(self) → int**: que retorna la cantidad de vuelos en pernocte que tenemos en pista.


```python
def validarTipo(variable:any, nombre:str, tipo:type)->any:
  if not isinstance(variable, tipo):
    raise Exception(f"La variable {nombre} debe ser de tipo {tipo}.")
  return variable

#Para poder hacer esto se deben usar punteros a funcion, sino con el tipo incorrecto la condicion se evalua antes
#de llamar a la funcion y va a fallar
def validarTipoCondicion(variable:any, nombre:str, tipo:type, condicion:bool = True):
  validarTipo(variable, nombre, tipo)
  if not condicion:
    raise Exception(f"La variable {nombre} no cumple la condición.")
  return variable

def validarNumeroPositivo(numero:int, nombre:str, tipo:type)->int:
  validarTipo(numero, nombre, tipo)
  if numero < 0:
    raise Exception(f"La variable {nombre} debe contener un numero positivo.")
  return numero
```

```python
import numpy as np

class Vuelo:
  def __init__(self,nombreDestino:str,matricula:str,cantHoras:int,origen:str):
    self.__nombreDestino = validarTipo(nombreDestino,"NombreDestino",str)
    self.__matricula = validarTipo(matricula,"matricula",str)
    self.__cantHoras = validarNumeroPositivo(cantHoras,"cantHoras",int)
    self.__origen = validarTipoCondicion(origen,"origen",str,("nacional","internacional"))

  def __str__(self) -> str:
    return f"{self.__nombreDestino} {self.__matricula} {self.__cantHoras} {self.__origen}"

  def superaLaCantidadDeHoras(self,horas:int) -> bool:
    return self.__cantHoras >= horas

class Pista:
  def __init__(self,cantFilas:int, cantPosFila:int):
    self.__cantFilas = validarNumeroPositivo(cantFilas,"cantFilas",int)
    self.__cantPosFila = validarNumeroPositivo(cantPosFila,"cantPosFila",int)
    self.__aeropuerto = np.empty((cantFilas,cantPosFila),Vuelo)

  def __str__(self) -> str:
    return f"{self.__aeropuerto}"

  def arriboDelVuelo(self, nombre:str, mat:str, cantHorasPos:int, origen:str):

    vuelo = Vuelo(nombre, mat, cantHorasPos, origen)
    vueloAsignado = False

    if(origen == "internacional"):
      for posicionFila in range(2):
        for posicionColumna in range(self.__cantPosFila):
          if(self.__aeropuerto[posicionFila][posicionColumna] == None and vueloAsignado == False): #Uso el vueloAsignado porque si estuviera la primera condicion sola el for sigue recorriendo
            self.__aeropuerto[posicionFila][posicionColumna] = vuelo                               #y seguiria encontrando posiciones con None por lo tanto asignando el vuelo en todas las posiciones
            vueloAsignado = True
    else:
      for posicionFila in range(2, self.__cantFilas):
        for posicionColumna in range(self.__cantPosFila):
          if(self.__aeropuerto[posicionFila][posicionColumna] == None and vueloAsignado == False):
            self.__aeropuerto[posicionFila][posicionColumna] = vuelo
            vueloAsignado = True

    if(vueloAsignado == False):
      raise Exception ("No hay posicion valida para asignar el vuelo")

  def esArriboPernocte(self,horas:int = 6) -> bool:
    vueloEncontrado = False
    for fila in self.__aeropuerto:
      for vuelo in fila:
        if(vuelo != None and vuelo.superaLaCantidadDeHoras(horas)):
          vueloEncontrado = True

    return vueloEncontrado

  def cantPernocte(self,horas:int = 6) -> int:
    cantidad = 0
    for fila in self.__aeropuerto:
      for vuelo in fila:
        if(vuelo != None and vuelo.superaLaCantidadDeHoras(horas)):
          cantidad += 1

    return cantidad
```

```python
pista1 = Pista(3,2) # Pista de 3 filas de 2 posiciones cada una.

pista1.arriboDelVuelo("FLN5950","kje",4,"internacional") #Ingresa en la posición (0,0), primera fila
pista1.arriboDelVuelo("GIG5900","kec",7,"internacional") # Ingresa en la posición (0,1), primera fila
pista1.arriboDelVuelo("MDZ5052","kdq",6,"nacional") # Ingresa en la posición (2,0), tercer fila
pista1.arriboDelVuelo("NQN5301","keh",2,"nacional") # Ingresa en la posición (2,1), tercer fila


print("\n", "Hay algún vuelo pernocte:", pista1.esArriboPernocte()) #-> True

print("\n", "Cantidad vuelos pernocte:", pista1.cantPernocte()) #-> 2
```
