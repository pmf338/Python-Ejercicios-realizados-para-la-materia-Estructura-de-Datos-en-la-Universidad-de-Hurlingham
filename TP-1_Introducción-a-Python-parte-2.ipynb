### **Ejercicio 3**

Escribir una función que calcule y retorne el factorial de un número natural. La operación factorial (!) se define de la siguiente manera:

image.png

```
def factorial(n:int) -> int:
    if n == 0:
        resultado = 1
    else:
      resultado = 1
    for operacionFactorial in range(n, 0, -1):
        resultado = resultado * operacionFactorial

    return resultado

print(f"Factorial de 5: {factorial(5)}")
print(f"Factorial de 0: {factorial(0)}")
```

### **Ejercicio 5**

Escribir la función máximo, que recibe 2 números por parámetro y retorna el mayor. Luego, usando esta función, escriba un programa que pida 10 números al usuario por teclado y al finalizar informe el mayor por pantalla.

```
def maximo(num1: int, num2: int) -> int:
    if num1 > num2:
        max = num1
    else:
        max = num2
    return max


def maximoPorIngreso():
    for ingresoNumeros in range(10):
        numero = int(input("ingrese un numero: "))
        if ingresoNumeros == 0:
            max = numero
        else:
            max = maximo(max, numero)
    return max


print(f"El numero maximo es: {maximoPorIngreso()}")
```

### **Ejercicio 6**

Escribir una función que calcule el área de un círculo y otra que calcule el volumen de un cilindro usando la primera función.

```
def areaCirculo(radio:int) -> float:
  pi = 3.14
  area = pi * radio**2
  return area

def volumenCilindro(radio:int,altura:int) -> float:
  volumen = areaCirculo(radio) * altura
  return volumen

print(f"El area del circulo es: {areaCirculo(5)}")
print(f"El volumen del cilindro es: {volumenCilindro(5,7)}")
```

Escribir una función que recibe tres números enteros por parámetros, calcula el promedio y retorna “APROBADO” si el promedio es mayor o igual a 7 o “DESAPROBADO” si es menor.
       

```
def promedio(nota1:int,nota2:int,nota3:int) -> str:
  promedio = (nota1 + nota2 + nota3) // 3
  if(promedio >= 7):
    return "APROBADO"
  else:
    return "DESAPROBADO"

print(f"La nota es: {promedio(6,7,8)}")
```

### **Ejercicio 9**

Escribir una función que dado un tiempo en horas, minutos y segundos retorne esa misma cantidad en segundos.

```
def horasASegundos(horas:int) -> int:
  minutos = horas * 60
  segundos = minutosASegundos(minutos)
  return segundos

def minutosASegundos(minutos:int) -> int:
  segundos = minutos * 60
  return segundos


def convertirASegundos(horas:int,minutos:int,seg:int) -> int:
  segundos = horasASegundos(horas) + minutosASegundos(minutos) + seg
  return segundos

print(f"El tiempo en segundos es: {convertirASegundos(2,45,35)}")
```

### **Ejercicio 10**

Escribir una función que dado un año, retorne verdadero si es bisiesto y falso en caso contrario.

Nota: Los años bisiestos son múltiplos de 4 y no son múltiplos de 100, pero los años múltiplos de 400 si son bisiestos.
Estos son algunos ejemplos de posibles respuestas: 2012 es bisiesto, 2010 no es bisiesto, 2000 es bisiesto, 1900 no es bisiesto

```
def bisiesto(anio: int) -> bool:
    if ((anio % 4 == 0) and (anio % 100 != 0 or anio % 400 == 0)):
        return True
    else:
        return False


print(bisiesto(2012))
print(bisiesto(2010))
```

### **Ejercicio 11**

Escribir un programa que pida dos años y escriba cuántos años bisiestos hay entre esas dos fechas (incluidos los dos años).

```
def bisiesto(anio: int) -> bool:
    if ((anio % 4 == 0) and (anio % 100 != 0 or anio % 400 == 0)):
        return True
    else:
        return False

def contarAniosBisiestos(anio1:int,anio2:int) -> int:
  contador = 0
  for anio in range(anio1, anio2 + 1):
        if bisiesto(anio):
            contador = contador + 1
  return contador

print(f"La cantidad de años bisiestos entre los dos años es: {contarAniosBisiestos(1556,2012)}")
```
