Escribir una función que recibe un vector de strings y retorna otro con los mismos elementos del vector de entrada pero en orden inverso.

```
import numpy as np

def invertirVector(vector: str) -> str:
    vectorInvertido = np.empty(len(vector), dtype=vector.dtype)
    pos = 0
    for elemento in reversed(vector):
        vectorInvertido[pos] = elemento
        pos += 1
    return vectorInvertido


vectorAInvertir = np.array(["cebolla", "banana", "papa", "manzana", "lechuga"])
print(f"{invertirVector(vectorAInvertir)}")
```

Escribir la función sumaDeVectores que recibe dos vectores de enteros del mismo tamaño y retorna otro con la suma de ambos. La suma de vectores se realiza componente a componente, segun la siguiente forma:

v<sub>1</sub> = [ v<sub>11</sub> , v<sub>12</sub> , v<sub>13</sub> , ... , v<sub>1N</sub> ]

v<sub>2</sub> = [ v<sub>21</sub> , v<sub>22</sub> , v<sub>23</sub> , ... , v<sub>2N</sub> ]

v<sub>1</sub> + v<sub>2</sub> = [ v<sub>11</sub> + v<sub>21</sub> , v<sub>12</sub> + v<sub>22</sub> , v<sub>13</sub> + v<sub>23</sub> , ... , v<sub>1N</sub> + v<sub>2N</sub> ]

```
import numpy as np

def sumaDeVectores(vec1:np.array,vec2:np.array) -> np.array:
  vectorSuma = np.zeros(len(vec1),int)
  for pos in range(len(vectorSuma)):
    vectorSuma[pos] = vec1[pos] + vec2[pos]
  return vectorSuma

vector1 = np.array([1,2,3,4,5])
vector2 = np.array([1,2,3,4,5])
print(sumaDeVectores(vector1,vector2))
```

### **Ejercicio 4**

Escribir una función que recibe un vector y desplaza todos sus elementos una posición hacia la derecha: el primero pasa a ser el segundo, el segundo pasa a ser el tercero y así sucesivamente. El último pasa a ser el primero.

Por ejemplo:

v1 = [2, 4, 1, 5, 7, **9**]

**desplazarADerecha(v1)**

=> v1 = [**9**, 2, 4, 1, 5, 7]

```
import numpy as np

def desplazarALaDerecha(vector:np.array) -> np.array:
  vector2 = vector.copy()
  for pos in range(len(vector)-1):
    vector[pos+1] = vector2[pos]
  vector[0] = vector2[len(vector)-1]
  return vector

vector = np.array([2,4,1,5,7,9])
print(desplazarALaDerecha(vector))
```

v1 = [2, 4, 1, 7, 6, 2]

```
import numpy as np

def desplazamientoPorIngreso(vector:np.array,posicion:int,elemento:int) -> np.array:
  vector2 = vector.copy()
  for pos in range(posicion,len(vector)-1):
    vector[pos+1] = vector2[pos]
  vector[posicion] = elemento
  return vector

vector = np.array([2, 4, 1, 7, 6, 2])
posicion = 2
elemento = 0
print(desplazamientoPorIngreso(vector,posicion,elemento))
```

### **Ejercicio 12**

Escribir una función que recibe las dimensiones y retorna una matriz rellena de la siguiente forma: En la posición ( n , m ) debe contener n + m.

Por ejemplo:

mat1 = rellenaMatriz(4,3)

=> mat1 =

| <!--   --> |Col 0|Col 1|Col 2|
|------------|:---:|:---:|:---:|
| **Fila 0** |  0  |  1  |  2  |
| **Fila 1** |  1  |  2  |  3  |
| **Fila 2** |  2  |  3  |  4  |
| **Fila 3** |  3  |  4  |  5  |

```
def rellenaMatriz(fila:int,col:int) -> np.array:
  matriz = np.zeros((fila,col),int)
  for posicionFila in range(len(matriz)):
    for posicionColumna in range(len(matriz[posicionFila])):
      matriz[posicionFila][posicionColumna] = posicionFila + posicionColumna
  return matriz

print(rellenaMatriz(4,3))
```

### **Ejercicio 13**

Escribir una función que recibe una matriz cuadrada (N x N) y calcula la suma de los elementos que están por encima de la diagonal principal (excluyendo la diagonal).

En el ejemplo de la imagen, la función deberia retornar: *b + c + f*

![texto alternativo](https://economipedia.com/wp-content/uploads/Captura-de-Pantalla-2019-07-29-a-les-13.18.05-1.png)

```
#nCol > nFila -> triangulo superior
#nCol == nFila -> diagonal pricipal
#nCol < nFila -> triangulo inferior
def sumaDeDiagonales(matriz:np.array) -> int:
  suma = 0
  nFilas,nCols = matriz.shape #3,3
  for posicionFila in range(nFilas):
    for posicionColumna in range(nCols):
      if(posicionColumna > posicionFila):
        suma += matriz[posicionFila][posicionColumna]
  return suma

matriz = np.array([[3,5,1],[2,8,14],[12,3,9],[23,4,0]])
print(sumaDeDiagonales(matriz))
```

### **Ejercicio 14**

Escribir una función que retorna *True* si una matriz cuadrada es matriz diagonal y *False* en caso contrario.

Una matriz cuadrada diagonal es una matriz que tiene ceros en todos sus elementos, menos en la diagonal principal (No importan los elementos de la diagonal principal, pueden ser cero o no).

**Si i, j son los índices de las filas y columnas de M**

**=> Una matriz M es diagonal si:**

**M [ i , j ] == 0 para todo i != j**

Matriz diagonal de 4 x 4:

| <!--   --> |Col 0|Col 1|Col 2|Col 3|
|------------|:---:|:---:|:---:|:---:|
| **Fila 0** |  0  |  0  |  0  |  0  |
| **Fila 1** |  0  |  2  |  0  |  0  |
| **Fila 2** |  0  |  0  |  4  |  0  |
| **Fila 3** |  0  |  0  |  0  |  0  |

```
def matrizDiagonal(matriz:np.array) -> bool:
  nFilas,nCols = matriz.shape
  for posicionFila in range(nFilas):
    for posicionColumna in range(nCols):
      if posicionFila != posicionFila and matriz[posicionFila][posicionColumna] != 0:
        return False
  return True

matriz = np.array([[1,0,0,0],[0,8,0,0],[0,0,9,0],[0,0,0,4]])
print(matrizDiagonal(matriz))
```

### **Ejercicio 15**

Escribir una función que retorna *True* si una matriz cuadrada de enteros es simétrica y *False* en caso contrario.

Una matriz simétrica es como la de la imagen (no importan los elementos de la diagonal principal):

![texto alternativo](https://economipedia.com/wp-content/uploads/Captura-de-Pantalla-2019-09-23-a-les-15.58.25.png)

Es decir:

**Si i, j son los índices de las filas y columnas de M**

**=> Una matriz M es simétrica si:**

**M [ i , j ] == M[ j, i ] para todo i != j**

Matriz simétrica de 4 x 4:

| <!--   --> |Col 0|Col 1|Col 2|Col 3|
|------------|:---:|:---:|:---:|:---:|
| **Fila 0** |  2  |  1  |  0  |  5  |
| **Fila 1** |  1  |  2  |  9  |  7  |
| **Fila 2** |  0  |  9  |  0  |  8  |
| **Fila 3** |  5  |  7  |  8  |  4  |

```
def matrizSimetrica(matriz:np.array) -> bool:
  nFila,nCol = matriz.shape
  for posicionFila in range(nFila):
    for posicionColumna in range(nCol):
      if matriz[posicionFila][posicionColumna] != matriz[posicionFila][posicionColumna]:
        return False
  return True

matriz = np.array([[2,1,0,5],[1,2,9,7],[0,9,0,8],[5,7,8,4]])
print(matrizSimetrica(matriz))
```

## Data analysis

### Subtask:
Perform detailed data analysis on the numerical features of the dataset.


**Reasoning**:
Calculate the correlation matrix, identify top correlated pairs, compute descriptive statistics, and detect outliers for numerical features.



```
import pandas as pd
import numpy as np

# Calculate the correlation matrix
correlation_matrix = df.corr()

# Identify top 5 most positively and negatively correlated pairs
positive_corr = correlation_matrix.unstack().sort_values(ascending=False)
positive_corr = positive_corr[positive_corr != 1]  # Exclude self-correlations
top_positive = positive_corr.head(5)

negative_corr = correlation_matrix.unstack().sort_values()
top_negative = negative_corr.head(5)

print("Top 5 Most Positively Correlated Feature Pairs:\n", top_positive)
print("\nTop 5 Most Negatively Correlated Feature Pairs:\n", top_negative)


# Descriptive statistics for each numerical feature
descriptive_stats = {}
for col in df.columns:
    descriptive_stats[col] = {
        'mean': df[col].mean(),
        'std': df[col].std(),
        'skewness': df[col].skew(),
        'kurtosis': df[col].kurt()
    }

print("\nDescriptive Statistics for Numerical Features:")
for col, stats in descriptive_stats.items():
    print(f"\nFeature: {col}")
    for stat_name, stat_value in stats.items():
        print(f"{stat_name}: {stat_value}")


# Outlier detection using IQR
outliers = {}
for col in df.columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    num_outliers = len(df[(df[col] < lower_bound) | (df[col] > upper_bound)])
    if num_outliers > 0:
        outliers[col] = num_outliers
        print(f"Feature '{col}' has {num_outliers} outliers.")
```
