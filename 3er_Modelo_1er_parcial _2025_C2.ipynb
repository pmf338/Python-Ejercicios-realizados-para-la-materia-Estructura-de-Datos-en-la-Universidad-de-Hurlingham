#Ejercicio 1

Escribir la función recursiva **parImpar(vector:np.array) → bool**, que dado un un arreglo “vector”, devuelve True si todos los elementos alternan entre par-impar (o impar-par). Las soluciones que NO sean recursivas no serán tomadas en cuenta.

Por ejemplo:

v1 = [4, 5, 10, 9, 8, 1, 12] => parImpar(v1) → True

v2 = [3, 10, 7, 9, 12, 15]  => parImpar(v2) → False

```python
import numpy as np

def primeroParSegundoImpar(vector:np.array) -> bool:
  return (vector[0] % 2 == 0 and vector[1] % 2 != 0)

def primeroImparSegundoPar(vector:np.array) -> bool:
  return (vector[0] % 2 != 0 and vector[1] % 2 == 0)

def parImpar(vector: np.array) -> bool:
  if(len(vector) <= 1):
    resultado = True
  elif(not(primeroParSegundoImpar(vector) or primeroImparSegundoPar(vector))):
    resultado = False
  else:
    resultado = parImpar(vector[1:])

  return resultado

vector1 =  [4, 5, 10, 9, 8, 1, 12]
vector2 = [3, 10, 7, 9, 12, 15]
print(parImpar(vector1))
print(parImpar(vector2))
```

#Ejercicio 2

El gobierno de un país cambia de idea y decide comenzar a atender la salud pública de sus habitantes, para esto decide desarrollar un software para optimizar el funcionamiento de los hospitales públicos. Implementar el **TDA Sala** que modela una sala de internación de un hospital, con **N** filas con **M** camas cada una. En cada cama se ubicará un **Paciente**, que se define con el DNI, la matrícula del médico que lo atiende y un diagnóstico, que puede ser: “grave” o “leve”.  
Al ubicar pacientes en la sala, tendremos en cuenta el diagnóstico, a pacientes “graves” los tenemos que ubicar cerca de las puertas de la sala (primera o última fila), mientras que los pacientes en estado “leve” solo pueden ir en el resto de la sala.  

Nota: Implementar el TDA Paciente con las operaciones necesarias. Pueden suponer que la cantidad de filas siempre es mayor o igual a 3 (al menos 2 filas de camas para graves y 1 para leves).

Implementar el TDA Sala que contenga al menos las siguientes operaciones:

* **__init__(self, cantFilas:int, cantCamas:int):** que recibe la cantidad de filas y cantidad de camas por fila y crea una variable de tipo Sala.

* **ingresarPaciente(self, dni:int, medico:int, diagnóstico:str):** que recibe los datos de un paciente y lo ubica en la sala. Arroja una excepción si el paciente no puede ser ubicado, por ejemplo porque es un paciente grave y los lugares correspondientes están llenos.

* **mismoMedico(self, filaN:int) → bool:** que devuelve True si en la fila “filaN” hay dos pacientes (o más) contiguos que los atienda el mismo médico.


```python
import numpy as np

def validarTipo(variable:any, nombre:str, tipo:type)->any:
  if not isinstance(variable, tipo):
    raise Exception(f"La variable {nombre} debe ser de tipo {tipo}.")
  return variable

def validarNumeroPositivo(numero:int, nombre:str, tipo:type)->int:
  validarTipo(numero, nombre, tipo)
  if numero < 0:
    raise Exception(f"La variable {nombre} debe contener un numero positivo.")
  return numero

def validarTipoCondicion(variable:any, nombre:str, tipo:type, condicion:bool = True):
  validarTipo(variable, nombre, tipo)
  if not condicion:
    raise Exception(f"La variable {nombre} no cumple la condición.")
  return variable

class Paciente:
    def __init__(self, dni: int, medico: int, diagnostico: str):
      self.__dni = validarNumeroPositivo(dni,"dni",int)
      self.__medico = validarNumeroPositivo(medico,"medico",int)
      self.__diagnostico = validarTipoCondicion(diagnostico,"diagnostico",str,("grave", "leve"))

    def __repr__(self) -> str:
      return f"{self.__dni} {self.__medico} {self.__diagnostico}"

    def getDiagnostico(self) -> str:
      return self.__medico

    def tieneMismoMedico(self, otro) -> bool:
      return self.__medico == otro.__medico

class Sala:
    def __init__(self, cantFilas: int, cantCamas: int):
        self.__cantFilas = cantFilas
        self.__cantCamas = cantCamas
        self.__camas = np.empty((cantFilas, cantCamas), Paciente)

    def __repr__(self) -> str:
        return f"{self.__camas}"

    def ingresarPaciente(self, dni: int, medico: int, diagnostico: str):
      paciente = Paciente(dni, medico, diagnostico)
      ingresado = False

      if(diagnostico == "grave"):
        for fila in [0, self.__cantFilas - 1]:
          for cama in range(self.__cantCamas):
            if(self.__camas[fila][cama] == None and ingresado == False):
              self.__camas[fila][cama] = paciente
              ingresado = True
      else:
        for fila in range(1, self.__cantFilas - 1):
          for cama in range(self.__cantCamas):
            if(self.__camas[fila][cama] == None and ingresado == False):
              self.__camas[fila][cama] = paciente
              ingresado = True

      if(ingresado == False):
        raise Exception("No hay lugar disponible para el paciente")

    def mismoMedico(self, filaN: int) -> bool:
      tienenMismoMedico = False

      for cama in range(self.__cantCamas - 1):
        paciente1 = self.__camas[filaN][cama]
        paciente2 = self.__camas[filaN][cama + 1]

        if(paciente1 != None and paciente2 != None):
            if(paciente1.tieneMismoMedico(paciente2)):
                tienenMismoMedico = True

      return tienenMismoMedico
```

```python
sala = Sala(10, 2) #Sala de 10 filas de 2 camas cada una

sala.ingresarPaciente(5648789, 154, "grave") #ingresa en la posicion (0,0) primera fila
sala.ingresarPaciente(6125479, 170, "leve") #ingresa en la posicion (1,0) segunda fila
sala.ingresarPaciente(10236854, 154, "grave") #ingresa en la posicion (0,1) primera fila
sala.ingresarPaciente(12478963, 170, "grave") #ingresa en la posicion (9,0), ultima fila
print(sala)

print(sala.mismoMedico(0)) #true
print(sala.mismoMedico(1)) #false


```
