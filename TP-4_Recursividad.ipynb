### **Ejercicio 1**

Implementar una función que calcule el factorial de un número de forma recursiva

```
def factorial(N:int)->int:
  fact = None
  if N == 0:
    fact = 1
  else:
    fact = N * factorial(N-1)
  return fact

print(factorial(5))
```

Luego escribir un programa que pida un número N (mayor o igual a 0) al usuario e imprima por pantalla los primeros N números de la serie de Fibonacci

Sobre la sucesion de Fibonacci: https://www.vix.com/es/btg/curiosidades/4461/que-es-la-sucesion-de-fibonacci




```
#Version recursiva
def fibo(N:int)->int:
  salida = None
  if N == 0:
    salida = 0
  elif N == 1:
    salida = 1
  else:
    salida = fibo(N-1) + fibo(N-2)
  return salida

print(fibo(5))
```

```
#Otra forma de verlo

def fibonacci(N:int)->int:
  salida = None
  if N <= 1:
    salida = N
  else:
    salida = fibonacci(N-2) + fibonacci(N-1)
  return salida

print(fibonacci(5))
```

```
#Version iterativa
def fiboIter(N:int)->int:
  salida = None

  n0 = 0
  n1 = 1

  if N == 0:
    salida = n0
  if N == 1:
    salida = n1
  if N > 1:
    for i in range(2, N+1):
      salida = n1 + n0
      n0 = n1
      n1 = salida
  return salida

print(fiboIter(5))
```

*repetirPalabra("hola",3) -> "holaholahola"*

```
def repetirPalabra(palabra:str,num:int) -> str:
  palabraRepetida = None
  if(num == 1):
    palabraRepetida = palabra
  else:
    palabraRepetida = repetirPalabra(palabra,num - 1) + palabra
  return palabraRepetida

print(repetirPalabra("hola",3))

```

```
#arreglo = [2,1,3,10,5,4,7]
#posicionDelMaximo(array, posInicio) = posInicio       si posInicio = len(array)-1
#                                    = posicion del max(array[0], array[posicion maximo en el resto])

def posicionDelMaximo(vector:np.array, posInicio:int=0)->int:
  posMaximo = None
  if posInicio == len(vector)-1:
    posMaximo = posInicio
  else:
    posicionMaximoResto = posicionDelMaximo(vector, posInicio+1)
    if vector[posInicio] >= vector[posicionMaximoResto]:
      posMaximo = posInicio
    else:
      posMaximo = posicionMaximoResto
  return posMaximo

arreglo = np.array([200,1,3,10,5,4,70])
print(posicionDelMaximo(arreglo))
```
