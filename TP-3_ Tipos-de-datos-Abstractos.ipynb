```
# validaciones antes de mi constructor
def validarAnio(anio:int)->int:
  validarTipo(anio, "anio", int)
  if anio < 1870:
    raise Exception(f"La variable año debe ser mayor a 1870")
  return anio

#Creo mi TDA
class Propiedad:
  def __init__(self, calle:str, numero:int, localidad:str, anio:int, cantAmb:int): #variables necesarias para crear una Propiedad.
      self.__calle = validarTipo(calle, "calle", str) # variables privadas
      self.__numero = validarNumeroPositivo(numero, "numero", int)
      self.__localidad = validarTipo(localidad, "localidad", str)
      self.__año = validarAnio(anio)
      self.__cantAmb = validarNumeroPositivo(cantAmb, "cantAmb", int)

  def __repr__(self)->str: # me va a devolver una Propiedad
    return f"{self.__calle} {self.__numero} ({self.__localidad})"

  def mismaLocalidad(propiedad1,propiedad2) -> bool:
    return propiedad1.__localidad == propiedad2.__localidad

  def mismaCalle(propiedad1,propiedad2) -> bool:
    return propiedad1.__calle == propiedad2.__calle

  def mayorNumeracion(propiedad1,propiedad2):
    if(Propiedad.mismaLocalidad(propiedad1,propiedad2) and Propiedad.mismaCalle(propiedad1,propiedad2)):
      if(propiedad1.__numero > propiedad2.__numero):
        return propiedad1
      else:
        return propiedad2
    else:
      raise Exception(f"Las propiedades no estan en la misma localidad")

  def impuestoAntiguo(propiedad) -> float:
    if(1 <= propiedad.__cantAmb and propiedad.__cantAmb <= 3):
        impuesto = 5
    elif(4 <= propiedad.__cantAmb and propiedad.__cantAmb <= 6):
        impuesto = 10
    elif(propiedad.__cantAmb > 6):
        impuesto = 25

  def impuestoActual(propiedad) -> float:
    if(1 <= propiedad.__cantAmb and propiedad.__cantAmb <= 5):
        impuesto = 5
    elif(propiedad.__cantAmb > 5):
        impuesto = 35

  def calcularimpuestoARBA(propiedad) -> float:
    impuesto = 0
    if(1871 <= propiedad.__año and propiedad.__año <= 1949):
      impuesto = Propiedad.impuestoAntiguo(propiedad)
    elif(propiedad.__año >= 1950):
      impuesto = Propiedad.impuestoActual(propiedad)
    return impuesto
```

```
def validarTipo(variable:any, nombre:str, tipo:type)->any:
  if not isinstance(variable, tipo):
    raise Exception(f"La variable {nombre} debe ser de tipo {tipo}.")
  return variable

#Para poder hacer esto se deben usar punteros a funcion, sino con el tipo incorrecto la condicion se evalua antes
#de llamar a la funcion y va a fallar
def validarTipoCondicion(variable:any, nombre:str, tipo:type, condicion:bool = True):
  validarTipo(variable, nombre, tipo)
  if not condicion:
    raise Exception(f"La variable {nombre} no cumple la condición.")
  return variable

def validarNumeroPositivo(numero:int, nombre:str, tipo:type)->int:
  validarTipo(numero, nombre, tipo)
  if numero < 0:
    raise Exception(f"La variable {nombre} debe contener un numero positivo.")
  return numero
```

```
def validarTipo(variable:any, nombre:str, tipo:type)->any:
  if not isinstance(variable, tipo):
    raise Exception(f"La variable {nombre} debe ser de tipo {tipo}.")
  return variable

def validarNumeroPositivo(numero:int, nombre:str, tipo:type)->int:
  validarTipo(numero, nombre, tipo)
  if numero < 0:
    raise Exception(f"La variable {nombre} debe contener un numero positivo.")
  return numero

class Cuenta:
    def __init__(self, numero: int, dni: str, saldo: float, interesAnual: float):
        self.__numero = validarNumeroPositivo(numero, "numero", int)
        self.__dni = validarTipo(dni, "dni", str)
        self.__saldo = validarNumeroPositivo(saldo, "saldo", (int, float))
        self.__interesAnual = validarNumeroPositivo(interesAnual, "interesAnual", (int, float))

    def __repr__(self) -> str:
        return f"{self.__numero} {self.__dni} (${self.__saldo})"

    def actualizarSaldo(self):
        interesDiario = self.__interesAnual / 365
        self.__saldo = self.__saldo + (self.__saldo * interesDiario / 100)

    def ingresarDinero(self, cantidad: float):
        self.__saldo = self.__saldo + cantidad

    def retirarDinero(self, cantidad: float):
        if cantidad >= self.__saldo:
            raise Exception(f"Saldo insuficiente")
        self.__saldo = self.__saldo - cantidad

cuenta1 = Cuenta(123, "30111222", 1000, 12)
print(cuenta1)

Cuenta.actualizarSaldo(cuenta1)
print(cuenta1)

Cuenta.ingresarDinero(cuenta1,5000)
print(cuenta1)

Cuenta.retirarDinero(cuenta1,5000)
print(cuenta1)
```

### **Ejercicio 3**

Implementar el TDA "Tiempo" que modela una duracion en horas, minutos y segundos.

Se deben implementar las siguientes operaciones:
- Constructor: Debe incluir las validaciones necesarias, la hora debe ser un número positivo y los minutos y segundos deben ser números positivos entre 0 y 59.
- \_\_repr__: Al usar la función *print* con una variable del tipo tiempo debe mostrar: **'horas':'minutos':'segundos'**.
- tiempoASegundos: Operación que toma una variable de tipo tiempo y retorna la cantidad en segundos.
- tiemposDesdeSegundos: Operación que recibe un tiempo en segundos como parámetro y retorna una variable de tipo tiempo, en horas minutos y segundos.
- mayorDuracion: Operación que recibe dos variables de tipo tiempo y retorna la de mayor duración.

```
class Tiempo:
    def __init__(self, horas: int, minutos: int, segundos: int):
        self.__horas = validarNumeroPositivo(horas, "horas", int)
        self.__minutos = validarTipoCondicion(minutos, "minutos", int, 0 <= minutos <= 59)
        self.__segundos = validarTipoCondicion(segundos, "segundos", int, 0 <= segundos <= 59)

    def __repr__(self) -> str:
        return f"{self.__horas}:{self.__minutos}:{self.__segundos}"

    def tiempoASegundos(self) -> int:
        return self.__horas * 3600 + self.__minutos * 60 + self.__segundos

    def tiempoDesdeSegundos(segundos: int) -> "Tiempo":
        horas = segundos // 3600
        minutos = (segundos % 3600) // 60
        segundosRestantes = segundos % 60
        return Tiempo(horas, minutos, segundosRestantes)

    def mayorDuracion(tiempo1, tiempo2) -> "Tiempo":
        if tiempo1.tiempoASegundos() > tiempo2.tiempoASegundos():
            return tiempo1
        else:
            return tiempo2

tiempo = Tiempo(5,25,42)
tiempo2 = Tiempo(7,32,40)
print(f"{Tiempo.tiempoASegundos(tiempo)}")
print(f"{Tiempo.tiemposDesdeSegundos(55)}")
print(f"{Tiempo.mayorDuracion(tiempo,tiempo2)}")
```

**Ejercicio 7**

Crear el TDA **“SalaDeCine”** que modela una sala de cine con los siguientes componentes: una estructura que permita almacenar las butacas, un elemento por cada butaca (por defecto será una sala de 30 filas, con 40 asientos cada una), el número de sala, el tipo (2 posibles: “2D” o “3D”). La sala se va ocupando con espectadores, un espectador se define por dos cosas, dni y edad.

Escribir para el TDA **SalaDeCine** al menos las siguientes operaciones:

*  **init(numero, tipo, cantFilas, cantAsientos)**: que construye una variable de tipo SalaDeCine
*  **ocuparAsiento(fila,columna,dni, edad)**: que recibe por parámetros la ubicación del espectador que ingresa (fila y columna) y ocupa la
butaca con el espectador. Si la butaca no existe o esta ocupada debe lanzar una excepción.
*  **asientosVaciosFila(fila)**: que recibe un número de fila y devuelve la cantidad de butacas vacías en esa fila.
*  **gananciaTotal()**: que calcula la ganancia total de la función, teniendo en cuenta que la entrada de los menores de 14 años sale \$2500 y la de los mayores \$4000 en las salas 2D y \$4000 y \$6000 en las salas 3D.
*  **vaciarSala()**: que vacía la sala luego de que termine la película, marcando como desocupada a todas las butacas para la próxima función.

```
### SALA DE CINE
# MATRIZ(30 X 40)
#N de sala
#tipo (2d o 3d) !!!!! VALIDACIONES

  #### Espectador
  # dni
  # edad

class Espectador:
    def __init__(self, dni: int, edad: int):
        self.__dni = validarTipo(dni, "dni", str)
        self.__edad = validarNumeroPositivo(edad, "edad", int)

    def __repr__(self) -> str:
        return f"DNI: {self.__dni} (edad {self.__edad})"

    def esMenor(self) -> bool:
        return self.__edad < 14


class SalaDeCine:
    def __init__(self, numero: int, tipo: str, cantFilas: 30, cantAsientos: 40):
        self.__numero = validarNumeroPositivo(numero, "numero", int)
        self.__tipo = validarTipoCondicion(tipo, "tipo", str, ("2D", "3D"))
        self.__cantFilas = validarNumeroPositivo(cantFilas, "cantFilas", int)
        self.__cantAsientos = validarNumeroPositivo(cantAsientos, "cantAsientos", int)
        self.__butacas = np.empty((cantFilas,cantAsientos),Espectador)

    def __repr__(self) -> str:
      return f"{self.__butacas}"

    def validarUbicacion(self, fila: int, columna: int):
        if not (fila < self.__cantFilas and columna < self.__cantAsientos):
            raise Exception(f"La butaca ({fila},{columna}) no existe.")

    def ocuparAsiento(self, fila: int, columna: int, dni: int, edad: int):
        self.__validarUbicacion(fila, columna)
        if (self.__butacas[fila][columna] != None):
            raise Exception(f"La butaca ({fila},{columna}) ya está ocupada.")
        self.__butacas[fila][columna] = Espectador(dni, edad)

    def asientosVaciosFila(self, fila: int) -> int:
        self.__validarUbicacion(fila, 0)
        vacios = 0
        for butaca in self.__butacas[fila]:
            if (butaca == None):
                vacios = vacios + 1
        return vacios

    def gananciaTotal(self) -> float:
        ganancia = 0
        if self.__tipo == "2D":
          precioMenor = 2500
        else:
          precioMenor = 4000
        if self.__tipo == "2D"
          precioMayor = 4000
        else:
          precioMayor = 6000

        for fila in self.__butacas:
            for butaca in fila:
                if (butaca != None):
                    if butaca.esMenor():
                        ganancia = ganancia + precioMenor
                    else:
                        ganancia = ganancia + precioMayor
        return ganancia

    def vaciarSala(self):
       filas, butacas = self.__butacas.shape
       self.__butacas = np.empty((filas,butacas),Espectador)
```

- **vaciarLaboraratorios()**: que elimina los laboratorios definidos

¡No son las únicas operaciones necesarias, agregar los métodos que crean necesarios, como el constructor!

```
import numpy as np

class Laboratorio:
  def __init__(self,computadoras:int,sillas:int,ventanas:int):
    self.__cantComputadoras = computadoras
    self.__cantSillas = sillas
    self.__cantVentanas = ventanas

  def __repr__(self) -> str:
    return f"{self.__cantComputadoras} {self.__cantSillas} {self.__cantVentanas}"

  def tieneMasVentanasQue(self, otro):
    return self.__cantVentanas > otro.__cantVentanas

  def tieneMasComputadorasQue(self, cantidad):
        return self.__cantComputadoras > cantidad

class EdificioLaboratorios:
  def __init__(self,cantLaboratorios:int):
    self.__laboratorios = np.empty(cantLaboratorios, Laboratorio)
    #self.__laboratorios.fill(None)

  def __repr__(self) -> str:
    return f"{self.__laboratorios}"

  def definirLaboratorio(self,indiceLaboratorio:int, laboratorio:Laboratorio):
    #validarIndiceLaboratorio()
    if(self.__laboratorios[indiceLaboratorio]) == None:
      self.__laboratorios[indiceLaboratorio] = laboratorio
    else:
      raise Exception(f"El laboratorio ya existe")

  def laboratorioConMasVentanas(self) -> Laboratorio:
    laboratorioConMasVentanas = self.__laboratorios[0]
    for elemento in self.__laboratorios:
        if elemento.tieneMasVentanasQue(laboratorioConMasVentanas):
          laboratorioConMasVentanas = elemento
    return laboratorioConMasVentanas

  def laboratoriosGrandes(self,nCompus:int) -> int:
    cont = 0
    for elemento in self.__laboratorios:
      if(elemento.tieneMasComputadorasQue(nCompus)):
        cont += 1
    return cont

  def vaciarLaboratorios(self):
      self.__laboratorios.fill(None)

laboratorio1 = Laboratorio(25,50,4)
laboratorio2 = Laboratorio(20,40,5)
laboratorio3 = Laboratorio(16,32,4)
laboratorio4 = Laboratorio(22,44,3)

edificioLaboratorios = EdificioLaboratorios(4)

edificioLaboratorios.definirLaboratorio(0,laboratorio1)
edificioLaboratorios.definirLaboratorio(2,laboratorio2)
edificioLaboratorios.definirLaboratorio(1,laboratorio3)
edificioLaboratorios.definirLaboratorio(3,laboratorio4)

print(edificioLaboratorios.laboratorioConMasVentanas())
print(edificioLaboratorios.laboratoriosGrandes(19))

```

Implementar el TDA **Examen** que contenga al menos las siguientes operaciones:

- **init**: que recibe la fecha, la cantidad de hileras y la cantidad de mesas por hilera del aula.
- **asignarLugar(nombre, dni)**:	que recibe los datos de un/a estudiante y la/o acomoda en el aula retornando la posición. Si el aula esta llena debe lanzar una excepción.
- **alumnosJovenes(dni)**: que cuenta y retorna la cantidad de estudiantes que poseen un numero de DNI mayor al que se recibe por parámetro.
- **empezarDeNuevo()**: que deja el aula sin asignar (vacía).

No son las únicas operaciones necesarias, agregar las operaciones y funciones que consideren necesarios, como por ejemplo __repr__.

```
import numpy as np

class Fecha:

  def __init__(self, dia:int, mes:int, anio:int):
    self.__dia = dia
    self.__mes = mes
    self.__anio = anio

  def __repr__(self):
    return f'{self.__dia}/{self.__mes}/{self.__anio}'


class Estudiante:

  def __init__(self, nombre:str, dni:int):
    self.__nombre = nombre
    self.__dni = dni

  def __repr__(self):
    return f'{self.__nombre}'

  def esJoven(self, dni:int)->bool:
    return self.__dni > dni


class Examen:

  def __init__(self, fecha:Fecha, nHileras:int, nMesas:int):
    self.__fecha = fecha
    self.__aula = np.empty((nHileras, nMesas), Estudiante)


  def __repr__(self):
    return f'{self.__aula}'


  #Subtarea de 1ra función
  def estaLibreInicio(self, fila:int)->bool:
    libre = True
    nFilas, nCols = self.__aula.shape

    if nCols > 1:
      libre = self.__aula[fila, 1] == None

    return libre


  #Subtarea de 1ra función
  def estaLibreFin(self, fila:int)->bool:
    libre = True
    nFilas, nCols = self.__aula.shape

    if nCols > 1:
      libre = self.__aula[fila, -2] == None
    return libre


  #Subtarea de 1ra función
  def estaLibreMedio(self, fila:int, col:int)->bool:
    return self.__aula[fila, col-1] == None and self.__aula[fila, col+1] == None


  #Subtarea de 1ra función
  def esLugarLibre(self, fila:int, col:int)->bool:
    lugarLibre = False
    nFilas, nCols = self.__aula.shape

    if col == 0:
      lugarLibre = self.estaLibreInicio(fila)
    elif col == nCols-1:
      lugarLibre = self.estaLibreFin(fila)
    else:
      lugarLibre = self.estaLibreMedio(fila, col)

    return lugarLibre


  #1ra función
  def asignarLugar(self, nombre:str, dni:int)->tuple[int, int]:
    filaUbicacion = colUbicacion = None
    asignado = False
    nFilas, nCols = self.__aula.shape

    nFila = 0

    while not asignado and nFila < nFilas:
      nCol = 0
      while not asignado and nCol < nCols:
        if self.__aula[nFila, nCol] == None and self.esLugarLibre(nFila, nCol):
          self.__aula[nFila, nCol] = Estudiante(nombre, dni)
          asignado = True
          filaUbicacion = nFila
          colUbicacion = nCol
        nCol += 1
      nFila += 1

    if not asignado:
      raise Exception("El aula se encuentra llena")

    return filaUbicacion, colUbicacion


  #2da función
  def alumnosJovenes(self, dni:int)->int:
    cantJovenes = 0

    for fila in self.__aula:
      for estudiante in fila:
        if estudiante != None and estudiante.esJoven(dni):
          cantJovenes += 1

    return cantJovenes


  #3ra funcion
  def empezarDeNuevo(self)->None:
    nFilas, nCols = self.__aula.shape
    self.__aula = np.empty((nFilas, nCols), Estudiante)
```

```
print("Ejemplo de cómo declarar un examen:")
examen = Examen(Fecha(22,9,2025),4,5)
print(examen)
print("\n")


print("Lugares asignados:")
print(examen.asignarLugar('Lucia',25741284))
print(examen.asignarLugar('Matias',21741284))
print(examen.asignarLugar('Violeta',34741284))
print(examen.asignarLugar('Diego',42741284))
print("\n", examen)
print("\n")


print("Cantidad de alumnos jovenes:", examen.alumnosJovenes(32000000))
print("\n")


print("Dejar aula vacia:")
examen.empezarDeNuevo()
print(examen)
```
